```python
# Importing the pandas library and giving alias
import pandas as pd  
```

```python
# Reading from a csv file and assigning it to a data frame object
df = pd.read_csv(r"C:\Users\nawaz\Desktop\python for data engineering\datasets\Flavors.csv")  
```

```python
# Grouping up the data by one column
group_by_frame = df.groupby('Base Flavor')
```

```python
# The grouped data must have an aggregate function to show results
group_by_frame
```

```python
# Note : mean() and median() functions could not be performed on string type columns
group_by_frame.describe()
```

```python
group_by_frame.sum()
```

```python
group_by_frame.count()
```

```python
# We can perform arbitrary number of aggregate functions on arbitrary number of columns as below :
group_by_frame.agg({'Base Flavor' : ['min', 'max', 'sum', 'count'], 
                               'Flavor Rating' : ['mean', 'median', 'sum', 'max']})
```

```python
# Grouping the data up by multiple columns
df.groupby(['Base Flavor', 'Liked']).max()
```
