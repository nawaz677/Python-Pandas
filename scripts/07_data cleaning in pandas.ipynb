```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\nawaz\Desktop\python for data engineering\datasets\Customer Call List.xlsx")
df
```

```python
df.drop_duplicates(inplace = True)
```

```python
df = df.drop(columns = 'Not_Useful_Column')
df
```

```python
df['Last_Name'] = df['Last_Name'].str.strip('/._')
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '', regex = True)
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')
```

```python
df[['Street_Address', 'State', 'Zip_Code']] = df['Address'].str.split(',', n = 2, expand = True)
```

```python
df['Paying Customer'] = df['Paying Customer'].str.replace('Y$', 'Yes', regex = True)
df['Paying Customer'] = df['Paying Customer'].str.replace('N$', 'No', regex = True)
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Y$', 'Yes', regex = True)
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('N$', 'No', regex = True)
```

```python
df = df.replace('N/a', '')
df = df.fillna('')
```

```python
for x in df.index:
    if df.loc[x, 'Do_Not_Contact'] == 'Yes':
        df.drop(index = x, inplace = True)
df
```

```python
for x in df.index:
    if df.loc[x, 'Phone_Number'] == '':
        df.drop(x, inplace = True)
        
# another way to drop row with null value in column(s)
# df = df.dropna(subset = 'Phone_Number', inplace = True)
df
```

```python
df = df.reset_index(drop = True)
```
