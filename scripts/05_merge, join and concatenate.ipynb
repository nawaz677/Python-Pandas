```python
import pandas as pd
```

```python
df1 = pd.read_csv(r"C:\Users\nawaz\Desktop\LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"C:\Users\nawaz\Desktop\LOTR 2.csv")
df2
```

```python
df1.merge(df2) # 'Inner' is the default
```

```python
# on keyword is used for checking matching columns. The columns in the on list will appear only once, rest if common will appear
# as duplicates. 
df1.merge(df2, how = 'outer', on = ['FellowshipID'])
```

```python
# If 'on' is omitted by default Pandas will use all common columns for matching and those columns will appear only once
df1.merge(df2, how = 'outer')
```

```python
# Exactly like left exclusive or right exclusive join in SQL
df1.merge(df2, how = 'right_anti', on = ['FellowshipID', 'FirstName'])
```

```python
df1.join(df2, how = 'outer', on = 'FellowshipID', lsuffix = '_left', rsuffix = '_right')
```

```python
# Best use join when working with indexes otherwise its complicated than merge and also involves more work
df3 = df1.set_index('FellowshipID')
df3
```

```python
df4 = df2.set_index('FellowshipID')
df4
```

```python
df3.join(df4, how = 'outer', lsuffix = '_left', rsuffix = '_right')
```

```python
df3.join(df4, how = 'outer', lsuffix = '_left', rsuffix = '_right')
```

```python
# axis = 0 will give the common columns for inner join
pd.concat([df3, df4], join = 'inner', axis = 0)
```

```python
# axis = 1 will give the common rows for inner join
pd.concat([df3, df4], join = 'inner', axis = 1)
```

```python
pd.concat([df3, df4], join = 'outer')
```

```python
# By default merge() is 'inner' and join() is 'left'.
# Use merge() when combining datasets using matching columns, such as joining customers and orders using CustomerID.
# Use join() when the data is already indexed meaningfully, or when you want to combine DataFrames using their indexes.
```
