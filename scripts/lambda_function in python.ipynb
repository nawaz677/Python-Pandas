```python
prices = ['$12.55', '$2.99', '$105']

list(map(lambda p: p.replace('$', ''), prices))
```

```python
students = [['umer', 90], 
            ['hasan', 59], 
            ['ali', 71]]

list(filter(lambda row: row[1] >= 70, students))
```
